```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

titanic = pd.read_csv(url)

# Display the first 5 rows
titanic.head()
```

```python
plt.hist(titanic['Age'].dropna(), bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
plt.scatter(titanic['Age'], titanic['Fare'], alpha=0.5, color='purple')
plt.title('Age vs. Ticket Fare')
plt.xlabel('Age')
plt.ylabel('Fare Paid (in Pounds)')
plt.show()
```

```python
sns.barplot(x='Pclass', y='Survived', data=titanic)
plt.title('Survival Rate by Pässenger Class')
plt.xlabel('Passenger Class (1 = 1st, 3 = 3rd) ')
plt.ylabel('Survival Rate (0 to 1) ')
plt.show()
```

```python
sns.set_theme(style="whitegrid")  # Clean white background with grid line
plt.figure(figsize=(10, 5))       # Make the picture wider (10 inches )

# Use a beautiful color palette (viridis)
sns.barplot(x='Pclass', y='Survived', data=titanic, palette='viridis')
plt.title('Survival Rate by Passenger Class',fontsize=16)
plt.xlabel('Passenger Class',fontsize=12)
plt.ylabel('Survival Rate',fontsize=12)
plt.show()
```

```python
sns.boxplot(x='Pclass', y='Fare', data=titanic)
plt.title('Fare Distribution by Class'),
plt.xlabel('Passenger Class')
plt.ylabel('Fare Paid')
plt.show( )
```

```python
plt.plot(titanic['Survived'].cumsum(), color='green')
plt. title('Cumulative Survivors Over Time')
plt. xlabel('Passenger ID [order on Ship)')
plt.ylabel('Total Survivors So Far')
plt. show()
```
